```python
rows19 = len(insp_19)
rows20 = len(insp_20)
rows21 = len(insp_21)
rows22 = len(insp_22)
rows23 = len(insp_23)

geometries_length = rows19 + rows20 + rows21 + rows22 + rows23
print(geometries_length)
```

```python
colnames_23 = {
    "system_created_at": "system_cre",
    "address_suite": "address_su",
    "address_thoroughfare": "address_th",
    "address_locality": "address_lo",
    "address_sub_admin_area": "address__2",
    "address_postal_code": "address_po",
    "address_full": "address_fu",
    "inspectiondate_calculate": "inspection",
    "structuretype": "structuret",
    "address_admin_area": "address_ad",
    "address_country": "address_co"
}

insp_23 = insp_23.rename(columns=colnames_23)
```

```python
insp_19["year"] = 2019
insp_20["year"] = 2020
insp_21["year"] = 2021
insp_22["year"] = 2022
insp_23["year"] = 2023
```

```python
len(insp_19.columns)
```

```python
len(insp_23.columns)
```

```python
inspections = pd.concat(
    [insp_19, insp_20, insp_21, insp_22, insp_23], axis=0, ignore_index=True
)
```

```python
inspections
```

```python
# CHUNK FOR BUFFER_GEOMETRIES; RUN IF USING BUFFER_GEOMETRIES, NOT TRAINING_GEOMETRIES

# converting 'system_cre' column to display single days, instead of days and times
inspections["system_cre"] = pd.to_datetime(inspections["system_cre"]).dt.date
```

```python
# CHUNK FOR BUFFER_GEOMETRIES; RUN IF USING BUFFER_GEOMETRIES, NOT TRAINING_GEOMETRIES

# Some dates in the 'inspection' column have the format '2023-08-22 00:00:00'.
# This chunk is to remove the occurrences of '00:00:00' and remove trailing whitespace.

inspections["inspection"] = (
    inspections["inspection"]
    .astype(str)
    .str.replace(" 00:00:00", "", regex=False)
    .str.strip()
)
```

```python
# Fixing CRS to 3310, Albers

inspections = inspections.to_crs(epsg=3310)

print(inspections.crs)
```

```python
# adding address_id to be unique for every unique inspection
inspections["inspection_id"] = np.arange(1, len(inspections) + 1)
```

```python
inspections.inspection_id
```

```python
inspections.to_file(
    "/capstone/wildfire_prep/data/PUZZLE_PIECES/inspections_master.geojson",
    driver="GeoJSON",
)
```
