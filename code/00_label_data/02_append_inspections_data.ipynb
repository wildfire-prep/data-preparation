```python
# LOADING TRAINING GEOMETRIES INSTEAD
insp_19 = gpd.read_file(
    "/capstone/wildfire_prep/data/training_geometries/training_geometries_2019.geojson"
)
insp_20 = gpd.read_file(
    "/capstone/wildfire_prep/data/training_geometries/training_geometries_2020.geojson"
)
insp_21 = gpd.read_file(
    "/capstone/wildfire_prep/data/training_geometries/training_geometries_2021.geojson"
)
insp_22 = gpd.read_file(
    "/capstone/wildfire_prep/data/training_geometries/training_geometries_2022.geojson"
)
insp_23 = gpd.read_file(
    "/capstone/wildfire_prep/data/training_geometries/training_geometries_2023.geojson"
)
```

```python
rows19 = len(insp_19)
rows20 = len(insp_20)
rows21 = len(insp_21)
rows22 = len(insp_22)
rows23 = len(insp_23)

geometries_length = rows19 + rows20 + rows21 + rows22 + rows23
print(geometries_length)
```
