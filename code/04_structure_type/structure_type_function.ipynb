```python
# Replace following information as needed

# Load one year of data - change filepath as needed
in23 = gpd.read_file(
    "/capstone/wildfire_prep/data/inspections_data/cleaned_status/inspections_2023.geojson"
)

# Load structuretype_codes.csv - metadata containing structure type and structure code numbers. Change filepath to location on your machine. 
structuretype_codes_meta = '/capstone/wildfire_prep/data/metadata/structuretype_codes.csv'

# Output filepath/output destination folder - change as needed. 
output_path = "/capstone/wildfire_prep/data/PUZZLE_PIECES"

# Output File Name - change as desired!
filename = 'data_with_structuretype_encoded.geojson'
```

```python
# View only status and structuretype columns, to see what it looks like before the function.
# After the function, the 'structuretype' column will be gone, and the 'structure_code' column will replace it 

in23[['status', 'structuretype']]
```

```python
def structuretype_codes(data):
    
    # Load structure codes from metadata folder: change path as needed
    codes = pd.read_csv(structuretype_codes_meta).rename(
        columns={"structure_type": "structuretype"}
    )

    # Merging structure_codes.csv metadata to the given dataset, dropping extra 'structuretype' column after merge
    data = data.merge(
        codes[["structuretype", "structure_code"]],
        on="structuretype",
        how="left",
    ).drop(columns=["structuretype"])
    
    return data
```

```python
create_codes = structuretype_codes(in23)

create_codes.to_file(
    f'{output_path}/{filename}',
    index=False,
)
```
